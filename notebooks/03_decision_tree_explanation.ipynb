# Decision tree explanation

Train a tree, inspect its rules, follow one sample through the tree, and export the report.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from ml_method_reports import report_for

X, y = make_classification(n_samples=150, n_features=4, n_informative=3, n_redundant=0, random_state=11)
feature_names = ["signal_a", "signal_b", "signal_c", "signal_d"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=11, stratify=y)
tree = DecisionTreeClassifier(max_depth=3, random_state=11).fit(X_train, y_train)
```

```python
print(export_text(tree, feature_names=feature_names))
```

```python
selected_sample_index = 0
tree.decision_path(X_test[[selected_sample_index]]).indices
```

```python
report_for(tree) \
    .with_training_data(X_train=X_train, y_train=y_train, feature_names=feature_names) \
    .with_test_data(X_test=X_test, y_test=y_test, feature_names=feature_names) \
    .with_options(selected_sample_index=selected_sample_index, dataset_source="synthetic tree dataset") \
    .display()
```

```python
report_for(tree) \
    .with_training_data(X_train=X_train, y_train=y_train, feature_names=feature_names) \
    .with_test_data(X_test=X_test, y_test=y_test, feature_names=feature_names) \
    .save("runtime/reports/decision_tree_notebook")
```
