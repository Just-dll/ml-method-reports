# Clustering exploration

Explore KMeans and Agglomerative Clustering. Cluster labels are group identifiers, not true class labels.

```python
import numpy as np
from sklearn.cluster import AgglomerativeClustering, KMeans
from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA
from ml_method_reports import report_for

X, true_groups = make_blobs(n_samples=180, centers=3, n_features=4, cluster_std=1.1, random_state=13)
feature_names = [f"feature_{i}" for i in range(X.shape[1])]

kmeans = KMeans(n_clusters=3, random_state=13, n_init=10).fit(X)
agglomerative = AgglomerativeClustering(n_clusters=3).fit(X)

projection = PCA(n_components=2, random_state=13).fit_transform(X)
print("PCA projection shape:", projection.shape)
print("KMeans cluster sizes:", dict(zip(*np.unique(kmeans.labels_, return_counts=True))))
print("KMeans centers shape:", kmeans.cluster_centers_.shape)
```

```python
report_for(kmeans).with_test_data(
    X_test=X, y_test=true_groups, feature_names=feature_names
).with_options(dataset_source="synthetic blob dataset", selected_sample_index=0).display()
```

```python
report_for(agglomerative).with_test_data(
    X_test=X, y_test=true_groups, feature_names=feature_names
).with_options(dataset_source="synthetic blob dataset").display()
```
