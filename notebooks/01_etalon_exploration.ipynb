# Etalon exploration

Explore a prototype classifier: class centers, distances, single-sample explanation, inline report, and export.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from ml_method_reports import EtalonClassifier, report_for

X, y = make_classification(n_samples=160, n_features=4, n_informative=3, n_redundant=0, random_state=2)
feature_names = ["length", "width", "density", "texture"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=2, stratify=y)

model = EtalonClassifier(metric="euclidean", normalization="standard").fit(X_train, y_train)
```

```python
model.get_class_centers()
```

```python
model.get_distances(X_test[:5])
```

```python
model.explain_prediction(X_test, y_test, sample_index=0)
```

```python
report_for(model) \
    .with_training_data(
        X_train=X_train,
        y_train=y_train,
        feature_names=feature_names,
    ) \
    .with_test_data(
        X_test=X_test,
        y_test=y_test,
        feature_names=feature_names,
    ) \
    .with_options(selected_sample_index=0, dataset_source="synthetic classification dataset") \
    .display()
```

```python
report_for(model) \
    .with_training_data(
        X_train=X_train,
        y_train=y_train,
        feature_names=feature_names,
    ) \
    .with_test_data(
        X_test=X_test,
        y_test=y_test,
        feature_names=feature_names,
    ) \
    .save("runtime/reports/etalon_notebook")
```
