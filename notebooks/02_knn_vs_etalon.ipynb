# KNN vs Etalon

Compare two metric methods on the same sample: etalon prototypes versus nearest training objects and voting.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from ml_method_reports import EtalonClassifier, report_for

X, y = make_classification(n_samples=180, n_features=4, n_informative=3, n_redundant=0, class_sep=1.4, random_state=5)
feature_names = [f"feature_{i}" for i in range(X.shape[1])]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=5, stratify=y)

etalon = EtalonClassifier().fit(X_train, y_train)
knn = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
selected_sample_index = 0
```

```python
print("Etalon explanation:", etalon.explain_prediction(X_test, y_test, selected_sample_index))
distances, neighbors = knn.kneighbors(X_test[[selected_sample_index]])
print("KNN neighbor indices:", neighbors[0])
print("KNN neighbor distances:", distances[0])
```

```python
report_for(etalon).with_data(
    X_train=X_train, X_test=X_test, y_train=y_train, y_test=y_test, feature_names=feature_names
).with_options(selected_sample_index=selected_sample_index, dataset_source="shared synthetic dataset").display()
```

```python
report_for(knn).with_data(
    X_train=X_train, X_test=X_test, y_train=y_train, y_test=y_test, feature_names=feature_names
).with_options(selected_sample_index=selected_sample_index, dataset_source="shared synthetic dataset").display()
```
